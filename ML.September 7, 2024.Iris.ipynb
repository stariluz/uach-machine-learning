### Librerias de codigo

```python
import numpy
from scipy.linalg import lu
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import Normalizer
```

### Cargar datos

```python
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data.head()
```

### Normalizar

```python
normalizer = Normalizer()
normalized_data = normalizer.fit_transform(data)
normalized_data_df = pd.DataFrame(normalized_data, columns=iris.feature_names)
normalized_data_df.head()
```

### Estandarizar

```python
standarizer = StandardScaler()
standarized_data = standarizer.fit_transform(data)
standarized_data_df = pd.DataFrame(standarized_data, columns=iris.feature_names)
standarized_data_df.head()
```
