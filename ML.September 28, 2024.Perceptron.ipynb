### Librerias de codigo

```python
import numpy
```

```python
# Definir la función de activación (función escalón)
def step_function(x):
    return numpy.where(x >= 0, 1, 0)

# Implementar el perceptrón
class Perceptron:
    def __init__(self, input_size, learning_rate=0.1):
        self.weights = numpy.zeros(input_size + 1)  # +1 para el bias
        self.learning_rate = learning_rate

    def predict(self, x):
        z = numpy.dot(x, self.weights[1:]) + self.weights[0]
        return step_function(z)

    def train(self, X, y, epochs=10):
        for _ in range(epochs):
            for i in range(len(X)):
                prediction = self.predict(X[i])
                error = y[i] - prediction
                self.weights[1:] += self.learning_rate * error * X[i]
                self.weights[0] += self.learning_rate * error

```

```python
# Inputs (Boolean table)
X = numpy.array([[0, 0], [0, 1], [1, 0], [1, 1]]) # Inputs
```

```python
# OR Perceptron
y_or = numpy.array([0, 1, 1, 1])  # Outputs

# Crear y entrenar el perceptrón
perceptron_or = Perceptron(input_size=2)
perceptron_or.train(X, y_or, epochs=100)
```

```python
# NAND Perceptron
y_nand = numpy.array([1, 1, 1, 0])  # Outputs

# Crear y entrenar el perceptrón
perceptron_nand = Perceptron(input_size=2)
perceptron_nand.train(X, y_nand, epochs=100)
```

```python
# AND Perceptron
y_and = numpy.array([0, 0, 0, 1])  # Outputs

# Crear y entrenar el perceptrón
perceptron_and = Perceptron(input_size=2)
perceptron_and.train(X, y_and, epochs=100)
```

```python
# Hacer predicciones
for x in X:
    prediction=perceptron_and.predict(
        [perceptron_or.predict(x), perceptron_nand.predict(x)]
    )
    print(f"{x} -> {prediction}")
```
