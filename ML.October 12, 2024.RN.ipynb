### Librerias de codigo

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical
```

```python
# Cargar el conjunto de datos de vino
data = pd.read_csv('./datasets/MentalHealthSurveyCleaned.csv')
```

```python
y = data.iloc[:, 1]
y.head()
```

```python
X = data.drop('age',axis=1)
X.head()
```

```python
y.unique()
```

```python
# Dividir los datos en conjunto de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=34)
```

```python
# Normalizar las características
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Las variables ya estan en formato categórico
# Convertir las etiquetas a formato categórico (one-hot encoding)
y_train -= 18
y_test -= 18
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
# Crear el modelo de red neuronal
model = Sequential()
# Añadir la primera capa densa con 64 neuronas y función de activación ReLU
model.add(Dense(96, activation='relu', input_shape=(X_train.shape[1],)))
# Añadir una segunda capa densa con 32 neuronas y función de activación ReLU
model.add(Dense(32, activation='relu'))
# Añadir la capa de salida con 8 neuronas (una por edad) y función de activación softmax
model.add(Dense(10, activation='softmax'))
```

```python
# Compilar el modelo
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
# Entrenar el modelo
history = model.fit(X_train, y_train, epochs=20, batch_size=8, validation_data=(X_test, y_test))
```

```python
# Evaluar el modelo en el conjunto de prueba
loss, accuracy = model.evaluate(X_test, y_test)
print(f'Precisión en el conjunto de prueba: {accuracy * 100:.2f}%')
```

```python

# Realizar predicciones
predictions = model.predict(X_test)
# Mostrar las primeras 5 predicciones junto con las etiquetas verdaderas
for i in range(5):
    print(f'Predicción: {np.argmax(predictions[i])}, Verdadero: {np.argmax(y_test[i])}')
```
